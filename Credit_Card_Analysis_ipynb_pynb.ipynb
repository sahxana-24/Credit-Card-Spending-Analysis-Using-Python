```python
city_spending = df.groupby("City")["Transaction_Amount"].sum()

plt.figure(figsize=(8,5))
plt.bar(city_spending.index, city_spending.values)

plt.title("Total Spending by City")
plt.xlabel("City")
plt.ylabel("Total Spending")

plt.show()
```

```python
card_spending = df.groupby("Card_Type")["Transaction_Amount"].sum()

plt.figure(figsize=(6,5))
plt.bar(card_spending.index, card_spending.values)

plt.title("Spending by Card Type")
plt.xlabel("Card Type")
plt.ylabel("Total Spending")

plt.show()
```

```python
merchant = df.groupby("Merchant_Category")["Transaction_Amount"].sum()

plt.figure(figsize=(10,5))
plt.bar(merchant.index, merchant.values)

plt.title("Merchant Category Spending")
plt.xlabel("Merchant Category")
plt.ylabel("Total Spending")

plt.xticks(rotation=45)

plt.show()
```

```python
payment = df["Payment_Mode"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(payment.values,
        labels=payment.index,
        autopct="%1.1f%%")

plt.title("Payment Mode Distribution")

plt.show()
```

```python
df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"])
```

```python
daily = df.groupby("Transaction_Date")["Transaction_Amount"].sum()

plt.figure(figsize=(12,5))

plt.plot(daily.index,
         daily.values)

plt.title("Daily Transaction Trend")

plt.xlabel("Date")

plt.ylabel("Amount")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Transaction_Amount"],
         bins=20)

plt.title("Transaction Amount Distribution")

plt.xlabel("Amount")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Age"],
            df["Transaction_Amount"])

plt.title("Age vs Transaction Amount")

plt.xlabel("Age")

plt.ylabel("Transaction Amount")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.boxplot(df["Transaction_Amount"])

plt.title("Transaction Amount Box Plot")

plt.ylabel("Amount")

plt.show()
```
